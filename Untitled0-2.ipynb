```python
import psutil

cpu_percent = psutil.cpu_percent(interval=1)
memory = psutil.virtual_memory().percent

if cpu_percent > 70:
    print("ALERT: CPU usage high!")
if memory > 80:
    print("ALERT: Memory usage high!")
```

```python
import smtplib

def send_email(alert_text):
    sender = "your_email@gmail.com"
    receiver = "receiver_email@gmail.com"
    password = "app_password"

    server = smtplib.SMTP('smtp.gmail.com', 587)
    server.starttls()
    server.login(sender, password)
    message = f"Subject: Alert\n\n{alert_text}"
    server.sendmail(sender, receiver, message)
    server.quit()

# Example usage
if cpu_percent > 70:
    send_email("CPU usage high!")
```

```python
import matplotlib.pyplot as plt

metrics = [cpu_percent, memory]
labels = ['CPU', 'Memory']

plt.bar(labels, metrics, color=['blue','green'])
plt.ylim(0,100)
plt.title("VM Resource Usage")
plt.show()
```
